### 3. Schéma du Processus Métier

Le processus métier de vente observé peut être représenté de la manière suivante, sans inventer d'étapes ni fixer le grain du futur fait : 

Événement principal : Vente (Commande) -> traçable via InvoiceNo.   
Produit vendu : -> traçable via StockCode et Description.   
Date de l'événement : -> traçable via InvoiceDate.  
Lieu géographique : -> traçable via Country.   Acteur (Client) : -> traçable via CustomerID (qu'il soit identifié ou inconnu).

```python
print("Quantités négatives :",
      (df["Quantity"] < 0).sum())

print("Prix unitaires négatifs :",
      (df["UnitPrice"] < 0).sum())

print("Date minimale :",
      df["InvoiceDate"].min())

print("Date maximale :",
      df["InvoiceDate"].max())
```

```python
questions = {
    "Evolution temporelle": [
        "InvoiceDate", "Quantity", "UnitPrice"
    ],
    "Analyse des produits": [
        "StockCode", "Description", "Quantity"
    ],
    "Analyse par pays": [
        "Country", "Quantity"
    ],
    "Commandes et clients": [
        "InvoiceNo", "CustomerID"
    ]
}

for question, champs in questions.items():
    disponibles = all(champ in df.columns for champ in champs)

    print(question, ":", "Faisable" if disponibles
          else "Champs manquants")

    print("Champs nécessaires :", champs)
```

### 4. Questions Décisionnelles et Faisabilité

**Question 1 (Évolution dans le temps) :** Comment évolue le volume des articles vendus mois par mois ?   
**Trace de faisabilité :** Nécessite InvoiceDate (axe temporel) et Quantity (mesure du volume).   
**Question 2 (Produits) :** Quels sont les produits générant le plus grand volume de commandes ?   
**Trace de faisabilité :** Nécessite StockCode / Description (axe produit) et Quantity (mesure).   
**Question 3 (Pays) :** Quelle est la répartition des ventes selon les différents pays d'expédition ?   
**Trace de faisabilité :** Nécessite Country (axe géographique) et Quantity ou le calcul de Quantity * UnitPrice.   
**Question 4 (Clients) :** Quelle est la proportion des commandes passées par des clients identifiés par rapport aux clients inconnus (sans identifiant) ?   
**Trace de faisabilité :** Nécessite CustomerID (pour filtrer les valeurs présentes ou manquantes) et InvoiceNo (pour compter les commandes).
